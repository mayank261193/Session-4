### Exercise 1 — Define a class

Create a class named `Car` that stores three attributes set in its `__init__` method: `make`, `model`, and `year`. No other methods are needed yet.

For example, `Car("Toyota", "Corolla", 2021)` should create a car whose `make` is `"Toyota"`, `model` is `"Corolla"`, and `year` is `2021`.

````python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

# Quick check
my_car = Car("Toyota", "Corolla", 2021)
print(my_car.make, my_car.model, my_car.year)
````

### Exercise 2 — Create objects and print attributes

Define a `Planet` class with two attributes, `name` and `number_of_moons`. Create three Planet objects — Earth (1 moon), Mars (2 moons), and Jupiter (95 moons) — and print each planet's name alongside its number of moons.

Expected output:
```
Earth has 1 moon(s)
Mars has 2 moon(s)
Jupiter has 95 moon(s)
```

````python
class Planet:
    def __init__(self, name, number_of_moons):
        self.name = name
        self.number_of_moons = number_of_moons

earth = Planet("Earth", 1)
mars = Planet("Mars", 2)
jupiter = Planet("Jupiter", 95)

for planet in [earth, mars, jupiter]:
    print(planet.name, "has", planet.number_of_moons, "moon(s)")
````

### Exercise 3 — A method that returns formatted info

Create a `Movie` class with attributes `title`, `director`, and `year`. Add a method called `details()` that **returns** (does not print) a single formatted string. Create one Movie object and print the result of calling `details()`.

Expected output:
```
'Inception' was directed by Christopher Nolan in 2010.
```

````python
class Movie:
    def __init__(self, title, director, year):
        self.title = title
        self.director = director
        self.year = year

    def details(self):
        return f"'{self.title}' was directed by {self.director} in {self.year}."

movie = Movie("Inception", "Christopher Nolan", 2010)
print(movie.details())
````

### Exercise 4 — A method that updates an attribute

Create a `SavingsAccount` class with attributes `owner` and `balance`. Add a method `deposit(amount)` that increases the balance by the given amount. Open an account for `"Neha"` with a balance of `5000`, deposit `1500`, then print the updated balance.

````python
class SavingsAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

account = SavingsAccount("Neha", 5000)
account.deposit(1500)
print(account.owner, "has a balance of", account.balance)
````

### Exercise 5 — Convert a dictionary into a class

The snippet below stores a laptop as a dictionary:

```python
laptop = {
    "brand": "Dell",
    "ram_gb": 16,
    "price": 72000,
}
```

Rewrite this as an equivalent class named `Laptop`. The `__init__` should set `brand`, `ram_gb`, and `price`, and add a method `specs()` that returns a formatted summary string. Create a Laptop object with the same values and print its specs.

````python
class Laptop:
    def __init__(self, brand, ram_gb, price):
        self.brand = brand
        self.ram_gb = ram_gb
        self.price = price

    def specs(self):
        return f"{self.brand} laptop | {self.ram_gb} GB RAM | Rs. {self.price}"

laptop = Laptop("Dell", 16, 72000)
print(laptop.specs())
````
