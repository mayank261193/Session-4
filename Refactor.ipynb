```python
#Refactor MP2 using classes
class ExpenseTracker:

    def __init__(self):
        self.expenses = []

    # Add Expense
    def add_expense(self):

        category = input("Enter category: ")
        amount = float(input("Enter amount: "))
        description = input("Enter description: ")

        expense = {
            "category": category,
            "amount": amount,
            "description": description
        }

        self.expenses.append(expense)

        print("Expense added successfully!")

    # View All Expenses
    def view_expenses(self):

        if not self.expenses:
            print("No expenses found.")
            return

        print("\n----- ALL EXPENSES -----")

        for i, expense in enumerate(self.expenses, start=1):

            print(
                f"{i}. "
                f"{expense['category']} | "
                f"Rs {round(expense['amount'], 2)} | "
                f"{expense['description']}"
            )

    # Total Spending
    def total_expenses(self):

        total = sum(expense["amount"] for expense in self.expenses)

        print(f"\nTotal spent: Rs {round(total, 2)}")

    # Total by Category
    def category_totals(self):

        totals = {}

        for expense in self.expenses:

            category = expense["category"]
            amount = expense["amount"]

            if category in totals:
                totals[category] += amount
            else:
                totals[category] = amount

        print("\n----- SPENDING BY CATEGORY -----")

        if not totals:
            print("No expenses found.")
            return

        for category, amount in totals.items():

            print(f"{category}: Rs {round(amount, 2)}")

    # Menu
    def menu(self):

        while True:

            print("\n==============================")
            print("     EXPENSE TRACKER")
            print("==============================")

            print("a). Add expense")
            print("b). View all expenses")
            print("c). View total spending")
            print("d). View spending by category")
            print("e). Quit")

            choice = input("\nEnter your choice: ")

            if choice == "a":
                self.add_expense()

            elif choice == "b":
                self.view_expenses()

            elif choice == "c":
                self.total_expenses()

            elif choice == "d":
                self.category_totals()

            elif choice == "e":
                print("Come again!")
                break

            else:
                print("Wrong choice. Please try again.")


# Create object
tracker = ExpenseTracker()

# Start program
tracker.menu()
```
